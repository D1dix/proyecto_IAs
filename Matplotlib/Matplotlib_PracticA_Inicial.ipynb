## Ejercicio 1: Interfaz Orientada a Objetos

**Objetivo:** Crear un gráfico de línea usando la interfaz OO de Matplotlib

El equipo de análisis necesita visualizar la evolución de temperatura media mensual de Madrid. El gráfico debe seguir las mejores prácticas: usar fig, ax = plt.subplots(), título descriptivo, etiquetas con unidades, grid y marcadores.



### Tareas

- Crear figura y axes con plt.subplots(figsize=(10, 6))
- Dibujar la línea con ax.plot() incluyendo marcadores
- Añadir título descriptivo con ax.set_title()
- Etiquetas de ejes con unidades (ej: 'Temperatura (°C)')
- Activar grid y usar tight_layout()


````python
# EJERCICIO 1: Interfaz Orientada a Objetos
# Crear un gráfico de línea con temperatura mensual

import matplotlib.pyplot as plt
import numpy as np

# Datos
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun',
         'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
temp_madrid = [6.3, 7.9, 11.2, 13.7, 17.4, 22.8,
               26.2, 25.6, 21.1, 15.2, 9.8, 6.9]

# Crear figura y axes con plt.subplots()
fig, ax = plt.subplots(figsize=(10, 6))

# Dibujar línea con marcadores
ax.plot(meses, temp_madrid, marker='o')

# Añadir título descriptivo
ax.set_title('Evolución de la temperatura media mensual en Madrid')

# Añadir etiquetas de ejes
ax.set_xlabel('Mes')
ax.set_ylabel('Temperatura (°C)')

# Añadir grid
ax.grid(True)

# Ajustar layout
plt.tight_layout()

plt.show()
````

## Ejercicio 1: Barras Agrupadas

**Objetivo:** Crear un gráfico de barras agrupadas comparando datos trimestrales

El departamento de ventas necesita comparar el rendimiento de 5 departamentos durante 3 trimestres. Debemos mostrar las 3 series de barras lado a lado con colores distinguibles y una leyenda clara.



### Tareas

- Calcular posiciones X usando np.arange()
- Definir ancho de barra (0.25) para 3 series
- Dibujar 3 series: x-ancho, x, x+ancho
- Configurar etiquetas del eje X con set_xticks/set_xticklabels
- Añadir leyenda con ax.legend()


````python
# EJERCICIO 2: Barras Agrupadas
# Comparar ventas trimestrales por departamento

import matplotlib.pyplot as plt
import numpy as np

# Datos
departamentos = ['Ventas', 'Marketing', 'I+D', 'Soporte', 'RRHH']
q1 = [320, 210, 450, 180, 95]
q2 = [380, 250, 470, 195, 100]
q3 = [410, 230, 520, 210, 98]

# Crear figura
fig, ax = plt.subplots(figsize=(10, 6))

# Calcular posiciones y ancho de barras
x = np.arange(len(departamentos))
ancho = 0.25

# Dibujar 3 series de barras
ax.bar(x - ancho, q1, width=ancho, label='Q1', color='#3498db')
ax.bar(x,         q2, width=ancho, label='Q2', color='#2ecc71')
ax.bar(x + ancho, q3, width=ancho, label='Q3', color='#e74c3c')

# Configurar etiquetas del eje X
ax.set_xticks(x)
ax.set_xticklabels(departamentos)

# Añadir título, etiquetas y leyenda
ax.set_title('Ventas trimestrales por departamento')
ax.set_xlabel('Departamento')
ax.set_ylabel('Ventas')
ax.legend()

ax.grid(True, axis='y')

plt.tight_layout()
plt.show()
````

## Ejercicio 2: Dashboard con Subplots

**Objetivo:** Construir un dashboard con 4 subplots (2x2) mostrando diferentes visualizaciones

El equipo de producto necesita un dashboard que muestre 4 perspectivas de los datos de una aplicación web: usuarios activos (línea), tiempos de sesión (histograma), páginas vs tiempo (scatter), y conversiones por fuente (barras).



### Tareas

- Crear figura con subplots(2, 2) y constrained_layout=True
- Subplot [0,0]: Gráfico de línea - usuarios activos
- Subplot [0,1]: Histograma - tiempos de sesión
- Subplot [1,0]: Scatter plot - páginas vs tiempo
- Subplot [1,1]: Barras - conversiones por fuente
- Añadir título general con fig.suptitle()


````python
# EJERCICIO 3: Dashboard con Subplots
# Crear dashboard 2x2 con diferentes visualizaciones

import matplotlib.pyplot as plt
import numpy as np

# Generar datos simulados
np.random.seed(42)

# Datos para cada gráfico
dias = np.arange(1, 31)
usuarios_activos = 1000 + np.cumsum(np.random.randn(30) * 50)

tiempos_sesion = np.random.exponential(5, 500)

paginas_vistas = np.random.poisson(8, 100)
tiempo_visita = paginas_vistas * 0.5 + np.random.randn(100) * 0.5

fuentes = ['Orgánico', 'Directo', 'Social', 'Email', 'Referral']
conversiones = [450, 320, 280, 180, 120]

# Crear figura con 4 subplots (2x2)
fig, ax = plt.subplots(2, 2, figsize=(14, 10), constrained_layout=True)

# Subplot (0,0): Gráfico de línea - Usuarios activos por día
ax[0, 0].plot(dias, usuarios_activos, marker='o')
ax[0, 0].set_title('Usuarios activos diarios')
ax[0, 0].set_xlabel('Día')
ax[0, 0].set_ylabel('Usuarios')
ax[0, 0].grid(True)

# Subplot (0,1): Histograma - Distribución tiempos de sesión
ax[0, 1].hist(tiempos_sesion, bins=30)
ax[0, 1].set_title('Distribución del tiempo de sesión')
ax[0, 1].set_xlabel('Tiempo (min)')
ax[0, 1].set_ylabel('Frecuencia')

# Subplot (1,0): Scatter - Páginas vistas vs Tiempo
ax[1, 0].scatter(paginas_vistas, tiempo_visita)
ax[1, 0].set_title('Páginas vistas vs Tiempo de visita')
ax[1, 0].set_xlabel('Páginas vistas')
ax[1, 0].set_ylabel('Tiempo de visita (min)')
ax[1, 0].grid(True)

# Subplot (1,1): Barras - Conversiones por fuente
ax[1, 1].bar(fuentes, conversiones)
ax[1, 1].set_title('Conversiones por fuente')
ax[1, 1].set_xlabel('Fuente')
ax[1, 1].set_ylabel('Conversiones')

# Título general del dashboard
fig.suptitle('Dashboard App Web',
             fontsize=18, fontweight='bold')

plt.show()
````

## Ejercicio 3: Anotaciones y Marcadores

**Objetivo:** Crear un gráfico con anotaciones avanzadas señalando puntos clave

Para un informe de análisis bursátil, necesitamos visualizar la evolución del precio de una acción con anotaciones que destaquen el máximo y mínimo del año, una línea con el precio medio, y una zona sombreada indicando un período de alta volatilidad.



### Tareas

- Dibujar la línea de precios
- Localizar máximo y mínimo con np.argmax/argmin
- Añadir anotaciones con flechas usando ax.annotate()
- Línea horizontal con precio medio usando ax.axhline()
- Zona sombreada con ax.axvspan()


````python
import matplotlib.pyplot as plt
import numpy as np

# Datos
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun',
         'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
ventas = [120, 135, 142, 155, 148, 162, 175, 168, 182, 195, 205, 220]

# Crear figura y axes
fig, ax = plt.subplots(figsize=(10, 6))

# Gráfico con estilo personalizado
ax.plot(
    meses,
    ventas,
    color='#1f77b4',
    marker='o',
    linestyle='-',
    linewidth=2,
    markersize=7
)

# Título y etiquetas con estilo
ax.set_title(
    'Evolución mensual de ventas',
    fontsize=14,
    fontweight='bold',
    pad=15
)
ax.set_xlabel('Mes', fontsize=11)
ax.set_ylabel('Ventas (miles €)', fontsize=11)

# Personalizar grid
ax.grid(
    True,
    linestyle='--',
    alpha=0.6
)

# Quitar bordes superiores y derechos (look más moderno)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Ajustes finales
plt.tight_layout()
plt.show()
````

# EJERCICIO 4: Anotaciones y Marcadores

import matplotlib.pyplot as plt
import numpy as np

# Generar datos de precio de accion
np.random.seed(42)
dias = np.arange(1, 253)
precio_base = 150
cambios = np.random.randn(252) * 2
precio = precio_base + np.cumsum(cambios)

# Crear figura
fig, ax = plt.subplots(figsize=(14, 7))

# Dibujar línea de precio
ax.plot(dias, precio, color='#2c3e50', linewidth=1.5, label='Precio ACME Corp')

# Encontrar máximo y mínimo
dia_max = np.argmax(precio)
precio_max = precio[dia_max]
dia_min = np.argmin(precio)
precio_min = precio[dia_min]

# Precio medio
precio_medio = np.mean(precio)

# Anotación para el máximo
ax.annotate(f'Máximo: {precio_max:.2f} EUR',
            xy=(dia_max, precio_max),
            xytext=(dia_max + 30, precio_max + 3),
            arrowprops=dict(arrowstyle='->', color='#27ae60', lw=2),
            fontsize=11, fontweight='bold', color='#27ae60')

# Anotación para el mínimo
ax.annotate(f'Mínimo: {precio_min:.2f} EUR',
            xy=(dia_min, precio_min),
            xytext=(dia_min + 30, precio_min - 5),
            arrowprops=dict(arrowstyle='->', color='#e74c3c', lw=2),
            fontsize=11, fontweight='bold', color='#e74c3c')

# Línea horizontal con precio medio
ax.axhline(y=precio_medio, color='#7f8c8d', linestyle='--',
           alpha=0.8, linewidth=2, label=f'Media: {precio_medio:.2f} EUR')

# Zona sombreada de alta volatilidad
ax.axvspan(60, 90, alpha=0.15, color='#e74c3c', label='Alta volatilidad')

# Puntos destacados
ax.scatter([dia_max], [precio_max], color='#27ae60', s=100, zorder=5)
ax.scatter([dia_min], [precio_min], color='#e74c3c', s=100, zorder=5)

# Título, etiquetas y leyenda
ax.set_title('Evolución del Precio ACME Corp - 2024',
             fontsize=16, fontweight='bold')
ax.set_xlabel('Dia de Trading', fontsize=12)
ax.set_ylabel('Precio (EUR)', fontsize=12)
ax.legend(loc='upper left', fontsize=10)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

## Ejercicio 4: Personalización de Estilo

**Objetivo:** Aplicar estilos personalizados y crear un gráfico profesional

Para una presentación ejecutiva, necesitamos un gráfico con estilo corporativo: colores específicos, fuentes consistentes, fondo personalizado, y elementos visuales pulidos como sombras y bordes redondeados.



### Tareas

- Configurar colores de fondo del figure y axes
- Personalizar colores de ejes, ticks y grid
- Usar una paleta de colores corporativa
- Añadir una caja de texto con estadísticas
- Eliminar spines innecesarios (top, right)


````python
# EJERCICIO 5: Personalización de Estilo
# Crear un gráfico con estilo corporativo profesional

import matplotlib.pyplot as plt
import numpy as np

# Datos de ventas mensuales
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun']
ventas_2023 = [45, 52, 48, 61, 55, 67]
ventas_2024 = [51, 58, 62, 70, 68, 78]

# Crear figura con tamaño apropiado
fig, ax = plt.subplots(figsize=(10, 6))

# Configurar colores de fondo (estilo corporativo)
fig.patch.set_facecolor('#f5f5f5')
ax.set_facecolor('#ffffff')

# Dibujar líneas con estilo corporativo
ax.plot(
    meses, ventas_2023,
    marker='o',
    linewidth=2,
    label='2023'
)

ax.plot(
    meses, ventas_2024,
    marker='o',
    linewidth=2,
    label='2024'
)

# Personalizar spines (eliminar top y right)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#cccccc')

# Personalizar grid
ax.grid(
    True,
    linestyle='--',
    alpha=0.5
)

# Calcular estadísticas
crecimiento_prom = np.mean(
    (np.array(ventas_2024) - np.array(ventas_2023)) / np.array(ventas_2023) * 100
)
max_venta = max(ventas_2024)

# Añadir caja de texto con estadísticas
texto_stats = (
    f'Crecimiento promedio: {crecimiento_prom:.1f}%\n'
    f'Máxima venta 2024: {max_venta}'
)

ax.text(
    0.02, 0.95,
    texto_stats,
    transform=ax.transAxes,
    fontsize=10,
    verticalalignment='top',
    bbox=dict(
        boxstyle='round',
        facecolor='#ecf0f1',
        edgecolor='gray'
    )
)

# Título y etiquetas con estilo
ax.set_title(
    'Comparativa de ventas mensuales 2023 vs 2024',
    fontsize=14,
    fontweight='bold',
    pad=15
)
ax.set_xlabel('Mes', fontsize=11)
ax.set_ylabel('Ventas (miles €)', fontsize=11)

# Leyenda
ax.legend()

plt.tight_layout()
plt.show()
````

## Ejercicio 5: Gráfico de Área Apilada

**Objetivo:** Crear un gráfico de área apilada con stackplot

El equipo de marketing necesita visualizar la evolución del tráfico web por canal (orgánico, directo, social, referral) durante 12 meses, mostrando tanto la contribución individual como el total acumulado.



### Tareas

- Usar ax.stackplot() para áreas apiladas
- Aplicar colores con transparencia
- Añadir leyenda en posición óptima
- Formatear eje Y con separador de miles
- Añadir línea de total opcional


````python
# EJERCICIO 6: Gráficos de Distribución
# Analizar distribución de salarios por departamento

import matplotlib.pyplot as plt
import numpy as np

# Generar datos de salarios simulados
np.random.seed(42)
salarios_it = np.random.normal(55000, 12000, 200)
salarios_ventas = np.random.normal(48000, 15000, 150)
salarios_marketing = np.random.normal(45000, 10000, 100)
salarios_rrhh = np.random.normal(42000, 8000, 80)

todos_salarios = np.concatenate([salarios_it, salarios_ventas,
                                  salarios_marketing, salarios_rrhh])

# Crear figura con 2 subplots horizontales
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# Subplot izquierdo: Histograma de todos los salarios
media = np.mean(todos_salarios)
mediana = np.median(todos_salarios)

ax[0].hist(todos_salarios, bins=30, density=True, alpha=0.7, color='steelblue')
ax[0].axvline(x=media, color='red', linestyle='--', label='Media')
ax[0].axvline(x=mediana, color='green', linestyle=':', label='Mediana')
ax[0].set_title('Distribución de salarios')
ax[0].set_xlabel('Salario (€)')
ax[0].set_ylabel('Frecuencia')
ax[0].legend()
ax[0].grid(True)

# Subplot derecho: Boxplot por departamento
ax[1].boxplot([salarios_it, salarios_ventas, salarios_marketing, salarios_rrhh],
              labels=['IT', 'Ventas', 'Marketing', 'RRHH'])
ax[1].set_title('Salarios por departamento')
ax[1].set_ylabel('Salario (€)')
ax[1].grid(True)


# Título general
fig.suptitle('Análisis de la distribución salarial por departamento', fontsize=16)

plt.tight_layout()
plt.show()
````

# EJERCICIO 7: Mapa de Calor con Imshow

import matplotlib.pyplot as plt
import numpy as np

# Crear matriz de correlación simulada
np.random.seed(42)
variables = ['Precio', 'Cantidad', 'Descuento', 'Margen', 'Satisfaccion']
n = len(variables)

# Generar matriz simétrica de correlaciones
matriz = np.eye(n)
for i in range(n):
    for j in range(i+1, n):
        corr = np.random.uniform(-0.8, 0.9)
        matriz[i, j] = corr
        matriz[j, i] = corr

# Crear figura
fig, ax = plt.subplots(figsize=(10, 8))

# Crear heatmap
im = ax.imshow(matriz, cmap='coolwarm', vmin=-1, vmax=1, aspect='auto')

# Añadir colorbar
cbar = ax.figure.colorbar(im, ax=ax, shrink=0.8)
cbar.ax.set_ylabel('Coeficiente de Correlación', rotation=-90, va='bottom', fontsize=11)

# Añadir anotaciones con los valores
for i in range(n):
    for j in range(n):
        valor = matriz[i, j]
        color = 'white' if abs(valor) > 0.5 else 'black'
        ax.text(j, i, f'{valor:.2f}', ha='center', va='center',
                color=color, fontsize=12, fontweight='bold')

# Configurar etiquetas de ejes
ax.set_xticks(np.arange(n))
ax.set_yticks(np.arange(n))
ax.set_xticklabels(variables, fontsize=11)
ax.set_yticklabels(variables, fontsize=11)

# Rotar etiquetas del eje X
plt.setp(ax.get_xticklabels(), rotation=45, ha='right', rotation_mode='anchor')

# Título
ax.set_title('Matriz de Correlación - Variables de Ventas',
             fontsize=16, fontweight='bold', pad=20)

# Añadir bordes a las celdas
for edge, spine in ax.spines.items():
    spine.set_visible(True)
    spine.set_color('#cccccc')

plt.tight_layout()
plt.show()

## Ejercicio 7: Informe Ejecutivo Completo

**Objetivo:** Crear un informe visual completo con múltiples gráficos y layout profesional

El CEO necesita un informe visual de una página que muestre los KPIs del trimestre: evolución de ingresos (línea), distribución por producto (donut), rendimiento por región (barras horizontales), y métricas clave en tarjetas. El diseño debe ser profesional y listo para presentación.



### Tareas

- Usar GridSpec para layout personalizado
- Crear tarjetas de KPI con ax.text()
- Gráfico de línea con área sombreada
- Gráfico donut (pie con wedgeprops)
- Barras horizontales con valores
- Título general y estilo corporativo consistente


````python
############################################################
# EJERCICIO 8: INFORME VISUAL COMPLETO
############################################################
# Objetivo: Crear un informe visual profesional combinando múltiples técnicas

# EJERCICIO 8: Informe Visual Completo
# Dashboard ejecutivo con múltiples visualizaciones

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.gridspec import GridSpec

# Datos
np.random.seed(42)
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun',
         'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
ingresos = [120, 135, 142, 155, 148, 162, 175, 168, 182, 195, 205, 220]
productos = ['Software', 'Servicios', 'Hardware', 'Soporte']
ventas_producto = [45, 30, 15, 10]
regiones = ['Norte', 'Sur', 'Este', 'Oeste']
ventas_region = [350, 280, 220, 180]

# KPIs
total_ingresos = sum(ingresos) * 1000
crecimiento = ((ingresos[-1] - ingresos[0]) / ingresos[0]) * 100
mejor_mes = meses[np.argmax(ingresos)]

# Crear figura con GridSpec
fig = plt.figure(figsize=(14, 10))
gs = GridSpec(4, 3, figure=fig)

# ──────────────
# Fila 0: KPIs
# ──────────────
kpi1 = fig.add_subplot(gs[0, 0])
kpi2 = fig.add_subplot(gs[0, 1])
kpi3 = fig.add_subplot(gs[0, 2])

for ax in [kpi1, kpi2, kpi3]:
    ax.axis('off')

kpi1.text(0.5, 0.5, f'Ingresos Totales\n€{total_ingresos:,.0f}',
          ha='center', va='center', fontsize=14, fontweight='bold')

kpi2.text(0.5, 0.5, f'Crecimiento\n{crecimiento:.1f}%',
          ha='center', va='center', fontsize=14, fontweight='bold')

kpi3.text(0.5, 0.5, f'Mejor Mes\n{mejor_mes}',
          ha='center', va='center', fontsize=14, fontweight='bold')

# ─────────────────────────
# Fila 1-2: Línea principal
# ─────────────────────────
ax_linea = fig.add_subplot(gs[1:3, :])
ax_linea.plot(meses, ingresos, marker='o')
ax_linea.set_title('Evolución mensual de ingresos')
ax_linea.set_xlabel('Mes')
ax_linea.set_ylabel('Ingresos (miles €)')
ax_linea.grid(True)

# ─────────────────────────
# Fila 3: Donut + Barras
# ─────────────────────────
ax_donut = fig.add_subplot(gs[3, 0])
ax_barras = fig.add_subplot(gs[3, 1:3])

# Donut de productos
ax_donut.pie(
    ventas_producto,
    labels=productos,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.4}
)
ax_donut.set_title('Ventas por producto')

# Barras por región
ax_barras.bar(regiones, ventas_region)
ax_barras.set_title('Ventas por región')
ax_barras.set_xlabel('Región')
ax_barras.set_ylabel('Ventas')
ax_barras.grid(True, axis='y')

# Título general
fig.suptitle('Informe Ejecutivo de Ventas', fontsize=18, fontweight='bold')

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
````

# EJERCICIO 8: Informe Ejecutivo Completo

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.gridspec import GridSpec

# Datos del informe
meses = ['Oct', 'Nov', 'Dic']
ingresos = [980, 1050, 1200]
objetivo = [950, 1000, 1100]

productos = ['Software', 'Servicios', 'Hardware', 'Soporte']
ventas_producto = [45, 30, 15, 10]

regiones = ['Norte', 'Sur', 'Este', 'Oeste']
ventas_region = [320, 280, 250, 190]

kpis = [
    ('€3.2M', 'Ingresos Q4', '#3182ce'),
    ('+18%', 'Crecimiento', '#48bb78'),
    ('2,450', 'Clientes', '#ed8936'),
    ('94%', 'Satisfacción', '#9f7aea')
]

# Colores corporativos
COLORS = {'primary': '#1a365d', 'secondary': '#2b6cb0', 'bg': '#f7fafc'}

# Crear figura con GridSpec
fig = plt.figure(figsize=(16, 12))
fig.patch.set_facecolor(COLORS['bg'])
gs = GridSpec(4, 3, figure=fig, height_ratios=[1, 1.5, 1.5, 2], hspace=0.35, wspace=0.3)

# KPI Cards (fila 0, pero dividida en 4 partes)
for idx, (valor, titulo, color) in enumerate(kpis):
    # Calcular posición: dividir las 3 columnas en 4 partes
    if idx < 2:
        ax_kpi = fig.add_subplot(gs[0, idx])
    else:
        ax_kpi = fig.add_subplot(gs[0, 2]) if idx == 2 else fig.add_subplot(gs[1, 2])

    ax_kpi = fig.add_subplot(gs[0, :])
    break  # Solo necesitamos crear el área una vez

# Recrear KPIs como una fila de subplots
gs_kpi = gs[0, :].subgridspec(1, 4, wspace=0.1)
for idx, (valor, titulo, color) in enumerate(kpis):
    ax_kpi = fig.add_subplot(gs_kpi[0, idx])
    ax_kpi.set_facecolor('white')
    ax_kpi.text(0.5, 0.65, valor, ha='center', va='center', fontsize=24,
                fontweight='bold', color=color, transform=ax_kpi.transAxes)
    ax_kpi.text(0.5, 0.25, titulo, ha='center', va='center', fontsize=11,
                color='#4a5568', transform=ax_kpi.transAxes)
    ax_kpi.set_xticks([])
    ax_kpi.set_yticks([])
    for spine in ax_kpi.spines.values():
        spine.set_color('#e2e8f0')

# Gráfico principal: Ingresos vs Objetivo
ax_main = fig.add_subplot(gs[1:3, :2])
ax_main.set_facecolor('white')
ax_main.plot(meses, ingresos, marker='o', linewidth=3, markersize=10,
             color=COLORS['secondary'], label='Ingresos')
ax_main.plot(meses, objetivo, marker='s', linewidth=2, markersize=8,
             color='#a0aec0', linestyle='--', label='Objetivo')
ax_main.fill_between(meses, ingresos, objetivo, alpha=0.1, color=COLORS['secondary'])

for i, (m, v) in enumerate(zip(meses, ingresos)):
    ax_main.annotate(f'€{v}K', (m, v), textcoords='offset points',
                     xytext=(0, 12), ha='center', fontsize=11, fontweight='bold',
                     color=COLORS['primary'])

ax_main.set_title('Evolución de Ingresos Mensuales', fontsize=14, fontweight='bold',
                  color=COLORS['primary'], pad=10)
ax_main.set_ylabel('Ingresos (miles EUR)', fontsize=11)
ax_main.legend(loc='upper left', fontsize=10)
ax_main.grid(True, alpha=0.3)
ax_main.spines['top'].set_visible(False)
ax_main.spines['right'].set_visible(False)

# Donut de productos
ax_donut = fig.add_subplot(gs[3, 0])
colors_pie = ['#3182ce', '#48bb78', '#ed8936', '#9f7aea']
wedges, texts, autotexts = ax_donut.pie(ventas_producto, labels=productos,
                                         autopct='%1.0f%%', colors=colors_pie,
                                         wedgeprops=dict(width=0.5, edgecolor='white'),
                                         pctdistance=0.75)
for autotext in autotexts:
    autotext.set_fontsize(9)
    autotext.set_fontweight('bold')
    autotext.set_color('white')

ax_donut.set_title('Ventas por Producto', fontsize=12, fontweight='bold',
                   color=COLORS['primary'])

# Barras de regiones
ax_bar = fig.add_subplot(gs[3, 1:])
ax_bar.set_facecolor('white')
bars = ax_bar.barh(regiones, ventas_region, color=colors_pie, edgecolor='white', height=0.6)

for bar, val in zip(bars, ventas_region):
    ax_bar.text(val + 5, bar.get_y() + bar.get_height()/2, f'{val}K',
                va='center', fontsize=10, fontweight='bold', color=COLORS['primary'])

ax_bar.set_title('Ventas por Región', fontsize=12, fontweight='bold',
                 color=COLORS['primary'])
ax_bar.set_xlabel('Ventas (miles EUR)', fontsize=10)
ax_bar.spines['top'].set_visible(False)
ax_bar.spines['right'].set_visible(False)
ax_bar.set_xlim(0, max(ventas_region) * 1.15)

# Título general
fig.suptitle('INFORME EJECUTIVO - Q4 2024',
             fontsize=20, fontweight='bold', color=COLORS['primary'], y=0.98)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
